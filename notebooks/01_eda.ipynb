# Almaty LRT: exploratory data analysis

Exploratory analysis of delay data from LRT corridors. The dataset is synthetic and is used only to demonstrate the pipeline.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data' / 'lrt_data.csv')
df.head()
```

```python
df.info()
df.describe(include='all').T
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(df['delay_minutes'], kde=True, ax=axes[0], color='#0f766e')
axes[0].set_title('Delay distribution')
sns.boxplot(data=df, x='is_peak_hour', y='delay_minutes', ax=axes[1], color='#f59e0b')
axes[1].set_title('Delay: peak and off-peak hours')
plt.tight_layout()
```

```python
numeric_columns = [
    'elevation_slope_deg', 'lane_isolation_score', 'turning_conflicts',
    'passenger_density', 'delay_lag_15m', 'delay_lag_30m', 'delay_minutes'
]
plt.figure(figsize=(10, 7))
sns.heatmap(df[numeric_columns].corr(), annot=True, cmap='vlag', center=0)
plt.title('Numeric feature correlations')
plt.tight_layout()
```
